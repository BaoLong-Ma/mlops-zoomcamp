```python
import numpy as np
import pandas as pd
```

```python
from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
```

```python
df_janvier = pd.read_parquet("yellow_tripdata_2021-01.parquet")
df_janvier
```

```python
df_fevrier = pd.read_parquet("yellow_tripdata_2021-02.parquet")
df_fevrier
```

```python
# question 1
nb_colonne = len(df_janvier.columns)
nb_colonne
```

```python
df_janvier['duration'] = (df_janvier['tpep_dropoff_datetime'] - df_janvier['tpep_pickup_datetime']).dt.total_seconds() / 60
df_fevrier['duration'] = (df_fevrier['tpep_dropoff_datetime'] - df_fevrier['tpep_pickup_datetime']).dt.total_seconds() / 60
```

```python
# question 2 
std_val = df_janvier['duration'].std()
std_val
```

```python
nb_record_before = len(df_janvier)

df_janvier = df_janvier[(df_janvier['duration'] >= 1) & (df_janvier['duration'] <=60)]
df_fevrier = df_fevrier[(df_fevrier['duration'] >= 1) & (df_fevrier['duration'] <=60)]

nb_record_after = len(df_janvier)
```

```python
# question 3 
nb_removed = (nb_record_after / nb_record_before)*100
nb_removed
```

```python
categorical  = ['PULocationID', 'DOLocationID']

df_janvier[categorical] = df_janvier[categorical].astype(str)
dict_janvier = df_janvier[categorical].to_dict(orient='records')
dict_janvier
```

```python
df_fevrier[categorical] = df_fevrier[categorical].astype(str)
dict_fevrier = df_fevrier[categorical].to_dict(orient='records')
dict_fevrier
```

```python
dv_janvier = DictVectorizer()
dv_fevrier = DictVectorizer()
```

```python
Xj_train = dv_janvier.fit_transform(dict_janvier)
```

```python
# question 4
Xj_train.shape[1]
```

```python
target = 'duration'
y_train = df_janvier[target].values

lr = LinearRegression()
lr.fit(Xj_train, y_train)
y_pred = lr.predict(Xj_train)
```

```python
# question 5
rmse = root_mean_squared_error(y_train, y_pred)
rmse
```

```python
Xf_val = dv_fevrier.fit_transform(dict_fevrier)

y_val = df_fevrier[target].values
yf_pred = lr.predict(Xf_val)
```

```python
# question 6
rmse_fev = root_mean_squared_error(y_val, yf_pred)
rmse_fev
```
